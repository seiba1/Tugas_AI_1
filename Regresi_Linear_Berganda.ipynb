Saya menggunakan data prediksi **harga rumah** berdasarkan:
- Luas bangunan (m²)
- Jumlah kamar tidur
- Jarak ke pusat kota (km)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data_rumah = {
    'luas_bangunan': [45,  60,  75,  90,  100, 120, 55,  80,  110, 130, 70,  95],
    'jml_kamar':     [2,   2,   3,   3,   4,   4,   2,   3,   4,   5,   3,   4 ],
    'jarak_kota_km': [15,  10,  8,   5,   3,   2,   12,  6,   4,   1,   9,   7 ],
    'harga_juta':    [280, 370, 460, 580, 700, 880, 310, 510, 730, 1100, 420, 640]
}

df_rumah = pd.DataFrame(data_rumah)
print("Dataset Harga Rumah:")
print(df_rumah.to_string(index=False))
```

```python
X_rumah = df_rumah[['luas_bangunan', 'jml_kamar', 'jarak_kota_km']]
y_rumah = df_rumah['harga_juta']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_rumah, y_rumah, test_size=0.25, random_state=7
)

model_berganda = LinearRegression()
model_berganda.fit(X_train_r, y_train_r)

print(f"Intercept (a)         : {model_berganda.intercept_:.2f}")
print(f"Koef luas_bangunan (b1): {model_berganda.coef_[0]:.2f}")
print(f"Koef jml_kamar     (b2): {model_berganda.coef_[1]:.2f}")
print(f"Koef jarak_kota    (b3): {model_berganda.coef_[2]:.2f}")
```

```python
y_pred_berganda = model_berganda.predict(X_test_r)

print("Perbandingan Aktual vs Prediksi:")
hasil_r = pd.DataFrame({'Harga Aktual (Juta)': y_test_r.values,
                         'Harga Prediksi (Juta)': y_pred_berganda.round(1)})
print(hasil_r.to_string(index=False))

print(f"\nEvaluasi Model Berganda:")
print(f"  MAE  : {mean_absolute_error(y_test_r, y_pred_berganda):.2f}")
print(f"  RMSE : {np.sqrt(mean_squared_error(y_test_r, y_pred_berganda)):.2f}")
print(f"  R²   : {r2_score(y_test_r, y_pred_berganda):.4f}")
```

```python
# Prediksi rumah baru
rumah_baru = [[85, 3, 4]]  # 85 m², 3 kamar, 4 km dari kota
prediksi_rumah = model_berganda.predict(rumah_baru)
print(f"Prediksi harga rumah (85m², 3 kamar, 4km dari kota): Rp {prediksi_rumah[0]:.1f} juta")

plt.figure(figsize=(7, 5))
plt.scatter(y_test_r, y_pred_berganda, color='mediumseagreen', edgecolors='k', s=80)
plt.plot([y_test_r.min(), y_test_r.max()],
         [y_test_r.min(), y_test_r.max()], 'r--', label='Prediksi Sempurna')
plt.xlabel('Harga Aktual (Juta Rupiah)')
plt.ylabel('Harga Prediksi (Juta Rupiah)')
plt.title('Regresi Berganda: Aktual vs Prediksi Harga Rumah')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Analisis :**
- **Koefisien b1 (luas_bangunan)** positif → setiap tambahan 1 m² luas bangunan, harga naik secara proporsional.
- **Koefisien b2 (jml_kamar)** positif → jumlah kamar yang lebih banyak meningkatkan harga rumah.
- **Koefisien b3 (jarak_kota)** negatif → semakin jauh dari pusat kota, harga semakin turun. Ini sesuai dengan logika pasar properti.
- Dengan **R² yang tinggi**, model ini mampu menjelaskan sebagian besar variasi harga berdasarkan ketiga fitur tersebut.
- Dibandingkan regresi sederhana, model berganda lebih realistis karena mempertimbangkan beberapa faktor sekaligus.
- **Catatan**: Pada data nyata, faktor seperti kondisi lingkungan, usia bangunan, dan fasilitas sekitar juga berpengaruh besar.
