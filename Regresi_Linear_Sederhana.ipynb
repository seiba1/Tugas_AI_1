Saya menggunakan data luas toko (m2) vs omzet harian (juta rupiah), semakin besar toko, semakin tinggi potensi omzet.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data_toko = {
    'luas_toko_m2': [20, 35, 50, 60, 75, 90, 100, 120, 140, 160],
    'omzet_juta':   [2.1, 3.4, 5.0, 6.2, 7.8, 9.5, 11.0, 13.2, 15.5, 17.8]
}

df_toko = pd.DataFrame(data_toko)
print("Dataset Toko:")
print(df_toko.to_string(index=False))
```

```python
X_toko = df_toko[['luas_toko_m2']]
y_toko = df_toko['omzet_juta']

X_train, X_test, y_train, y_test = train_test_split(
    X_toko, y_toko, test_size=0.2, random_state=0
)

model_linear = LinearRegression()
model_linear.fit(X_train, y_train)

print(f"Intercept (a) : {model_linear.intercept_:.4f}")
print(f"Koefisien (b) : {model_linear.coef_[0]:.4f}")
print(f"\nPersamaan regresi: Omzet = {model_linear.intercept_:.2f} + {model_linear.coef_[0]:.2f} × Luas_Toko")
```

```python
y_pred_linear = model_linear.predict(X_test)

print("Perbandingan Aktual vs Prediksi:")
hasil = pd.DataFrame({'Luas (m²)': X_test.values.flatten(),
                      'Omzet Aktual': y_test.values,
                      'Omzet Prediksi': y_pred_linear.round(2)})
print(hasil.to_string(index=False))

print(f"\nEvaluasi Model:")
print(f"  MAE  : {mean_absolute_error(y_test, y_pred_linear):.4f}")
print(f"  MSE  : {mean_squared_error(y_test, y_pred_linear):.4f}")
print(f"  RMSE : {np.sqrt(mean_squared_error(y_test, y_pred_linear)):.4f}")
print(f"  R²   : {r2_score(y_test, y_pred_linear):.4f}")
```

```python
luas_baru = [[200]]
prediksi_baru = model_linear.predict(luas_baru)
print(f"Prediksi omzet toko seluas 200 m²: Rp {prediksi_baru[0]:.2f} juta/hari")

plt.figure(figsize=(8, 5))
plt.scatter(X_toko, y_toko, color='steelblue', label='Data Asli', zorder=5)
plt.plot(X_toko, model_linear.predict(X_toko), color='tomato', label='Garis Regresi')
plt.xlabel('Luas Toko (m²)')
plt.ylabel('Omzet Harian (Juta Rupiah)')
plt.title('Regresi Linear Sederhana: Luas Toko vs Omzet')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**ANALISIS:**

- **Intercept (a)** bernilai negatif kecil, artinya ketika luas toko = 0 m² maka omzet pun mendekati nol (masuk akal secara bisnis).
- Koefisien (b) ≈ 0.11, artinya setiap penambahan 1 m² luas toko, omzet meningkat sekitar Rp 110.000/hari.
- **Nilai R² mendekati 1.0** menunjukkan bahwa model sangat baik — luas toko mampu menjelaskan hampir semua variasi omzet dalam data ini.
- **RMSE** yang kecil menandakan bahwa error prediksi rata-rata masih dalam batas wajar.
- Hubungan antara luas toko dan omzet bersifat linear positif yang kuat, sehingga model ini cocok digunakan.
- **Keterbatasan**: Model ini hanya mempertimbangkan satu faktor (luas toko), padahal omzet nyata juga dipengaruhi oleh lokasi, promosi, jumlah produk, dll.
